```python
from numpy.linalg import norm


variants_prep = []

counts = Counter()

for var in db_people:
    counts["total"] += 1
    display_name = var["display_name"]
    display_norm = normalise_text(display_name)

    if display_norm.isalnum():
        last = display_norm
        first = ""
        counts["single"] += 1
    elif ", " in display_norm:
        counts["comma"] += 1
        parts = display_norm.split(", ")
        if len(parts) == 2:
            last = parts[0]
            first = parts[1]
        else:
            last = parts[0]
            first = parts[:-1]
    elif " " in display_norm:
        counts["space"] += 1
        parts = display_norm.rsplit(" ", 1)
        if len(parts) == 2:
            last = parts[1]
            first = parts[0]
        else:
            last = parts[-1]
            first = parts[0:-1]
    else:
        last = display_norm
        first = ""
        counts["single"] += 1

    variants_prep.append({
        "display_norm": display_norm,
        "last": last,
        "first": first,
        **var
    })

# with open("variants_prepped.json", "w") as f:
#     json.dump(variants_prep, f, ensure_ascii=False, indent=2)

    # rp(f"""
    # name: {name}
    # parts: {parts}
    # last: {last}
    # first: {first}
    # """)
rp(counts)
```
