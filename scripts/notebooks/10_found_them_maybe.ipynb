```python
missing_file = Path(project_root / "data_reload/for_reparsing/missing_entries.json")

with open(missing_file, "r") as f:
   missing_entries = json.load(f)



batched_folder = Path(project_root / "data_reload/for_reparsing/batched")


i = 0
total = math.ceil(len(missing_entries) / 25)
total_str = str(total)
rp(total)

for batch in batched(missing_entries, 25):
    i += 1
    batch_id = str(i).zfill(2)
    filename = "reparse_missing_" + batch_id + "-" + total_str + ".json"
    file_path = batched_folder / filename

    with open(file_path, "w") as f:
        json.dump(batch, f, ensure_ascii=False, indent=2)
    rp(f"file {i} of {total} saved")
```
